```python
#Import dependencies
import pandas as pd
from sklearn.model_selection import train_test_split, cross_val_score, KFold
from sklearn.linear_model import LinearRegression
```

```python
#bring in file
spotify=pd.read_excel('/content/low_popularity_spotify_data.csv.xlsx')
spotify.head()
```

```python
# call dummies(converting categorical column to dummies)
dummies = pd.get_dummies(spotify[['playlist_name','playlist_genre','playlist_subgenre']], drop_first=True)
dummies.head()
```

```python
# call numerical column, so i work with model ready features
spotify_num = spotify.select_dtypes(exclude='object')
spotify_num.head()
```

```python
# COncat both the dummies and numerical column
spotify_concat=pd.concat([dummies,spotify_num],axis=1)
spotify_concat.head()
```

The error message `ValueError: Input X contains NaN` indicates that the Linear Regression model cannot handle missing values in the input features. We need to handle these missing values before training the model. A common approach is to impute them with the mean of their respective columns.

```python
# Check for missing values in spotify_concat
print("Number of missing values per column before imputation:")
print(spotify_concat.isnull().sum()[spotify_concat.isnull().sum() > 0])
```

```python
# Impute missing values with the mean of each column
for col in spotify_concat.columns:
    if spotify_concat[col].isnull().any():
        spotify_concat[col] = spotify_concat[col].fillna(spotify_concat[col].mean())

print("Number of missing values per column after imputation:")
print(spotify_concat.isnull().sum()[spotify_concat.isnull().sum() > 0])
```

```python
# to call train and test data
X=spotify_concat.drop(['track_popularity'], axis=1)
y=spotify_concat['track_popularity']
```

```python
#split data
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=8)
```

```python

#Perform Cross Validation
Kf=KFold(n_splits=5, shuffle=True, random_state=7)
```

```python
# Instantiate LinReg

linreg=LinearRegression()
linreg_cv=cross_val_score(linreg, X_train, y_train, cv=Kf, scoring='neg_mean_squared_error')
linreg_cv
```
